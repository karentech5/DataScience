### Data Integration in Feature Engineering

### 1. What is Data Integration?

Data Integration is the process of combining data from multiple sources into a unified dataset for analysis or machine learning.

In feature engineering, integration helps create richer features by merging different datasets.

### Example:

- Customer dataset
- Transaction dataset
- Demographic dataset

All combined to build better ML features.

### 2. Why Data Integration is Important

Reasons:
 - Improves model accuracy
 - More information for prediction
 - Creates new features
 - Derived insights from multiple datasets
 - Removes data silos
 - Combines scattered information
 - Provides complete context

Example: customer + behavior + demographics

### 3. Common Data Sources

Data can come from many systems:
Source	LIKE: Databases, APIs, Files, Logs, External datasets, Streaming (Kafka, IoT sensors)

Example in banking:

Dataset	Features: 

Customer table	age, gender

Account table	balance

Transaction table	amount, frequency

### 4. Types of Data Integration
4.1 Schema Integration

Combine datasets with different structures.

Example:

Dataset A 

CustomerID | Name | Age

Dataset B

CustID | Address

Need to map:

CustomerID = CustID

4.2 Entity Resolution

Identify same entities across datasets.

Example:

John Smith
J. Smith
Smith, John

All may represent same customer.

Methods:
- String matching
- Fuzzy matching
- Record linkage

4.3 Data Fusion

Combine data from multiple sources into a single value.

Example:

Weather data:

Source A: 28°C
Source B: 27°C

Final value:

Average = 27.5°C
4.4 Data Aggregation

Summarize detailed data.

Example:

Transactions → Features

Feature	        | Meaning

TotalSpend	    |   Sum of transactions

AvgTransaction	|   Average transaction

TxnCount	    |   Number of transactions


### 5. Data Integration Techniques
5.1 Joins

Common SQL join types

5.2 Concatenation

Combine datasets with same structure.

Example:

January sales + February sales

Python:

import pandas as pd

df = pd.concat([jan_data, feb_data])
5.3 Key-Based Integration

Merge datasets using primary keys.

Example:

CustomerID | AccountID | ProductID

Python example:

df = pd.merge(customers, transactions, on="CustomerID")

### 6. Data Integration Challenges
6.1 Data Inconsistency

Example:

Dataset A:
    Gender = Male

Dataset B:
    Gender = M

Need standardization.

6.2 Duplicate Records

Example:

CustomerID = 1001
CustomerID = 1001

Must remove duplicates.

6.3 Missing Data

Example:

Age = NULL

Solutions:
- Imputation
- Dropping rows
- Filling default values

6.4 Different Units

Example:

Dataset A:  Price = USD

Dataset B:  Price = NPR

Need currency conversion.
