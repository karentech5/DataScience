###  test with head tail ,shape ,info describe

### Use these Pandas commands to test any dataset with head, tail, shape, info, and describe:

```python
import pandas as pd

# Load dataset (change file name/path)
df = pd.read_csv("datasets/Titanic-Dataset.csv")

# 1. First 5 rows
print("HEAD:")
print(df.head())

# 2. Last 5 rows
print("\nTAIL:")
print(df.tail())

# 3. Shape (rows, columns)
print("\nSHAPE:")
print(df.shape)

# 4. Info (columns, data types, null values)
print("\nINFO:")
df.info()

# 5. Statistical summary
print("\nDESCRIBE:")
print(df.describe())
```

What each does:

- df.head() → shows top 5 records
- df.tail() → shows last 5 records
- df.shape → shows (number_of_rows, number_of_columns)
- df.info() → shows column names, data types, missing values
- df.describe() → shows mean, min, max, std (numeric columns)
