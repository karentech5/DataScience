```python
### .loc — Label-based Selection (VERY IMPORTANT)
#df.loc[row_label, column_label]
```

Select a single row by label

```python
df.loc[0]
```

Select multiple rows

```python
df.loc[0:4]
#Returns rows from label 0 to 3 (inclusive).
```

Select specific column

```python
df.loc[:, 'Age']
# All rows, only the 'Age' column.
```

Select specific rows and columns

```python
df.loc[0:2, ['Name', 'Age']]
# Rows 0–2 and only Name & Age columns.
```

```python
df.loc[0:4]
```

```python
df.loc[0:4, ["Name", "Age", "Sex"]] #Select rows + columns
```

Boolean condition (very common)

```python
df.loc[df['Age'] > 30] #All rows where Age is greater than 30.
```

Condition + specific column

```python
df.loc[df['Age'] > 30, 'Name'] #Names of people whose age > 30.
```

Update values using .loc

```python
df.loc[df['Age'] > 30, 'Age'] = 40
#Changes Age to 40 where Age > 30.
```

Important notes

.loc uses labels (column names, index names)

Last value in slicing is included

```python
df.loc[0:2]   # includes 0,1,2
```

Difference from .iloc

.loc → label-based

.iloc → position-based

```python
df.loc[3, 'Name']   # by label
df.iloc[0, 0]       # by position
```

```python
df.loc[3, ["Name","Age"]] #Select a single value (cell)
```

```python
# Select all rows but specific columns
df.loc[:, ["Survived", "Pclass", "Fare"]]
```

### .iloc — Position-based Selection

df.iloc[row_position, column_position]

```python
df.iloc[0:5] #Select first 5 rows
```

```python
df
```

```python
df.iloc[0:5, 0:3] # Select rows & columns by position
```
