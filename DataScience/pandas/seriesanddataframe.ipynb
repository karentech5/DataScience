```python
# your task generate the another data using the list

import pandas as pd

# Original data using lists
names = ["Ram", "Shyam", "Hari", "Sita"]
ages = [23, 25, 22, 24]
cities = ["Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur"]

# Create DataFrame
df = pd.DataFrame({
    "Name": names,
    "Age": ages,
    "City": cities
})

print("Original Data:")
print(df)
```

### Generate another data column from existing list

Create a new column called "Age_after_5_years"

```python
df["Age_after_5_years"] = [age + 5 for age in ages]

print("\nUpdated Data:")
print(df)
```

Generate a completely new dataset using list

```python
# New list data
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]
prices = [800, 20, 35, 150]
quantity = [5, 50, 30, 10]

df2 = pd.DataFrame({
    "Product": products,
    "Price": prices,
    "Quantity": quantity
})

# Create new calculated column
df2["Total_Value"] = df2["Price"] * df2["Quantity"]

print("\nNew Generated Dataset:")
print(df2)
```

Detailed Comparison: Pandas Series vs DataFrame (with Code Examples)
🔹 1) Definition

Series

- A one-dimensional labeled array.
- Like a single column in Excel.
- Holds one type of data (numbers, text, etc.).

DataFrame

- A two-dimensional labeled data structure.
- Like a full Excel table.
- Contains multiple columns (multiple Series combined).

Basic

```python
import pandas as pd

# Series Example
s = pd.Series([10, 20, 30, 40])
print(s)

# DataFrame Example
df = pd.DataFrame({
    "Name": ["Ram", "Shyam", "Hari"],
    "Age": [23, 25, 22]
})
print(df)
```

3) Structure Difference

Feature	Series	DataFrame

Dimension	1D	2D

Looks like	Single column	Table

Components	Values + Index	Rows + Columns

Multiple columns	❌ No	✅ Yes

4) Data Representation

Series

```python
s = pd.Series([100, 200, 300], index=["a", "b", "c"])
print(s)
```

DataFrame

```python
df = pd.DataFrame({
    "Product": ["Pen", "Book", "Bag"],
    "Price": [10, 50, 500]
})
print(df)
```

5) How They Are Related

A DataFrame is a collection of Series.

Each column in a DataFrame is a Series.

```python
print(df["Product"])   # This returns a Series
```

6) Data Selection

Series Selection

```python
s = pd.Series([10, 20, 30], index=["x", "y", "z"])
print(s["y"])
```

DataFrame Selection

```python
df = pd.DataFrame({
    "Name": ["Ram", "Hari"],
    "Age": [23, 22]
})

print(df["Name"])     # Column selection
print(df.loc[0])      # Row selection
```

7) Creation Methods

Series from list

```python
s = pd.Series([1, 2, 3, 4])
```

DataFrame from dictionary

```python
df = pd.DataFrame({
    "A": [1, 2],
    "B": [3, 4]
})
```

8) Operations Comparison

Series Math Operation

```python
s = pd.Series([1, 2, 3])
print(s * 2)
```

DataFrame Math Operation

```python
df = pd.DataFrame({
    "A": [1, 2],
    "B": [3, 4]
})

print(df * 2)
```

9) Shape Difference

```python
print(s.shape)   # (3,)
print(df.shape)  # (2, 2)
```

Series → only length

DataFrame → rows × columns

🔹 10) Use Case Difference

Use Series when:

- Working with single column data
- Labels + values needed

Example: temperature list, marks list

Use DataFrame when:

- Working with full dataset
- Multiple columns

Like Excel table / database table

🔹 11) Memory Usage

- Series uses less memory
- DataFrame uses more memory (multiple Series combined)

12) Convert Between Them

Series → DataFrame

```python
s = pd.Series([10, 20, 30])
df = s.to_frame(name="Numbers")
print(df)
```

DataFrame → Series

```python
print(df["Numbers"])
```

13) Key Summary Table

Feature	Series	DataFrame

Dimension	1D	2D

Equivalent to	Single column	Full table

Index	Yes	Yes

Columns	No	Yes

Multiple data types	Limited	Yes

Part of	Base object	Collection of Series


14) Real-Life Example

Series

```python
marks = pd.Series([80, 75, 90], index=["Ram", "Shyam", "Hari"])
```

DataFrame

```python
students = pd.DataFrame({
    "Name": ["Ram", "Shyam", "Hari"],
    "Marks": [80, 75, 90],
    "City": ["KTm", "PKR", "LTP"]
})
print(students)
```

🔹 Final Conclusion

Series = Single column

DataFrame = Multiple columns (table)

Every column in DataFrame is a Series

Series is simpler; DataFrame is more powerful and widely used in data science.
