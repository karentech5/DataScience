```python
# the total number of elements of the array. 
a.size
```

```python
# an object describing the type of the elements in the array. 
a.dtype
```

```python
# the size in bytes of each element of the array. 
a.itemsize
```

```python
# the buffer containing the actual elements of the array. Normally, we won’t need to use 
a.data
```

### Initialize a Python NumPy Array Using Special Functions
NumPy provides several built-in functions to generate arrays with specific properties.

- np.zeros(): Creates an array filled with zeros.
- np.ones(): Creates an array filled with ones.
- np.full(): Creates an array filled with a specified value.
- np.arange(): Creates an array with values that are evenly spaced within a given range.
- np.linspace(): Creates an array with values that are evenly spaced over a specified interval.

```python
# np.arange
np.arange(1,11,2)
```

```python
# create a array using a arrange
np.arange(1,11,2)
```

### np.ceil() → Ceiling

Definition: Rounds each number UP to the nearest integer.

Always goes toward +∞, no matter what.

```python
import numpy as np

arr = np.array([1.2, 2.7, -1.5, -2.3])
arr
```

```python
np.ceil(arr)
```

### np.floor() → Floor

Definition: Rounds each number DOWN to the nearest integer.

Always goes toward -∞.

```python
np.floor(arr)
```

### np.round() → Round (to nearest)

Definition: Rounds each number to the nearest integer.

If decimal = 0.5 → rounds to nearest even integer (NumPy behavior).

```python
np.round(arr)
```
